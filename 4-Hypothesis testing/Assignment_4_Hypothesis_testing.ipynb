```python
import pandas as pd
import numpy as np
import scipy.stats as stats

## H0 --> There is no significant association between the type of device purchased and the customer satisfaction level
## Ha --> There is significant association between the type of device purchased and the customer satisfaction level
```

```python
Thermostat=[50,80,60,30,20]
Smartlight=[70,100,90,50,50]
alpha=0.05
dof=5-1
dof
```

```python
contingency_table = np.array([Thermostat, Smartlight])
contingency_table
```

```python
chi2_stat,p_value, dof, expected_val = stats.chi2_contingency(contingency_table)
print('chi_square statistics:',chi2_stat)
print('p_value:',p_value)
print('degree of freedom:',dof)
print('expected values:',expected_val)
```

```python
chi_critical = stats.chi2.ppf(0.95,4)
chi_critical
```
