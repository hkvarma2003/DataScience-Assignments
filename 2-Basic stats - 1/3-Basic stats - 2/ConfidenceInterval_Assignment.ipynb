A total of 15 print-heads were randomly selected and tested until failure. The durability of each print-head (in millions of characters) was recorded as follows:
1.13, 1.55, 1.43, 0.92, 1.25, 1.36, 1.32, 0.85, 1.07, 1.48, 1.20, 1.33, 1.18, 1.22, 1.29

Assignment Tasks
a. Build 99% Confidence Interval Using Sample Standard Deviation
Assuming the sample is representative of the population, construct a 99% confidence interval for the mean number of characters printed before the print-head fails using the sample standard deviation. Explain the steps you take and the rationale behind using the t-distribution for this task.

```python
import numpy as np
import scipy.stats as stats
from statsmodels.stats.weightstats import ztest

print_head=[1.13, 1.55, 1.43, 0.92, 1.25, 1.36, 1.32, 0.85, 1.07, 1.48, 1.20, 1.33, 1.18, 1.22, 1.29]
sample = len(print_head)   ## already given in the problem
alpha=0.005
total_area=1
degree_freedom=sample-1
sample =15

x_bar=sum(print_head)/len(print_head)
print(x_bar)
```

```python
sample_std=np.std(print_head)
sample_std
```

```python
t_critical=stats.t.ppf(total_area-0.005,degree_freedom)
```

```python
t_critical
```

```python
pop_mean_upper=x_bar+(t_critical*sample_std/np.sqrt(sample))
pop_mean_upper

pop_mean_lower=x_bar-(t_critical*sample_std/np.sqrt(sample))
pop_mean_lower
print(pop_mean_lower,pop_mean_upper)
## results are in Million characters
```

Reasons to choose T_test to calculate the Confidence Intervals:
1. Population Standard Deviation is not given
2.  Sample size is less than < 30

Build 99% Confidence Interval Using Known Population Standard Deviation
If it were known that the population standard deviation is 0.2 million characters, construct a 99% confidence interval for the mean number of characters printed before failure.


```python
sample =15
pop_std=0.2
x_bar=sum(print_head)/len(print_head)
print(x_bar)
```

```python
z_critical = stats.norm.ppf(total_area-0.005)
print(z_critical)
```

```python
confidence_interval_lower=x_bar-(z_critical*pop_std/np.sqrt(sample))
confidence_interval_upper=x_bar+(z_critical*pop_std/np.sqrt(sample))
print(confidence_interval_lower,confidence_interval_upper)
```
