```python
## this is a function to replace the Outliers with lowerbound and upper bound values, this technique called Capping

def outlier_capping(df,column):
  Q1=df[column].quantile(0.25)
  Q3=df[column].quantile(0.75)
  IQR=Q3-Q1
  lower_bound=Q1-1.5*IQR
  upper_bound=Q3+1.5*IQR
  df[column]=df[column].apply(lambda x:lower_bound if x<lower_bound else upper_bound if x>upper_bound else x)

## Calling outlier_Capping method for each column which are only Integer and Float datatypes
for col in df.select_dtypes(include=['int','float']).columns:
  outlier_capping(df,col)
```

```python
## After capping technique checking whether the dataset has outlier or not by using box plot.

df.boxplot()
```

```python
## Zscore Capping is another method to replace the outliers in the dataset

## but observed that zscore capping is not working, need to check the reason for not replacing the Lower and uppper bounderies. Please verify and suggest why the outlier are not removed in this technique.

def zscore_capping(df, column):
  lower_bound=df[column].mean()-(2*df[column].std())
  upper_bound=df[column].mean()+(2*df[column].std())
  df[column]=df[column].apply(lambda x:lower_bound if x<lower_bound else upper_bound if x>upper_bound else x)
  # print(df[column])

for col in df.select_dtypes(include=['int','float']).columns:
  zscore_capping(df,col)
```

```python
## VIF

from statsmodels.stats.outliers_influence import variance_inflation_factor

vif=pd.DataFrame()
y=df['LB']
x=df.drop(columns=['LB','DS','DP'])

x.columns
```

```python
##  VIF --> variance Inflation factor technique to find the corelation between feature to feature.

## VIF > 10 the column needs to be dropped, but if it is important we need to take decision whether the column to be dropped or not.

vif['columns']=x.columns

vif['VIF']=[variance_inflation_factor(x.values,i) for i in range(len(x.columns))]
vif.sort_values(by='VIF',ascending=False)
```

```python
## F-Regression is the technique to find the relationship between feature and target.
## Target is LB and features are other than LB column

from sklearn.feature_selection import f_regression

f_reg=f_regression(x,y)
f_reg

```
