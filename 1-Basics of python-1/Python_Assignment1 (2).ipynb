```python
def checkPrime(num):            ## Function that counts the number of Multipliers
  counter = 0
  for i in range(1, num):
    if(num%i ==0):
      counter += 1
  return counter

num = int(input(print('Enter a number:')))    ## User input


counter = checkPrime(num)       ## Calling the function and find the number of multipliers

if(counter > 2):                ## check the number is prime or not based on counter returned by Function
  print(num, 'is not a prime number')
else:
  print(num, 'is a prime number')
```

```python
def sumEvenNums():        ##  Function to find Sum of Even Numbers between 100 to 200
  sumEvens =0
  for i in range(100,201):
    if(i%2==0):
      sumEvens += i**2
  return sumEvens

def sumOddNums():        ##  Function to find Sum of Odd Numbers between 100 to 200
  sumOdds = 0
  for i in range(100,201):
    if(i%2!=0):
      sumOdds += i**2
  return sumOdds
option = input(print('Enter your option, sum of Even or Odd:'))
if(option.lower() == 'even'):
  print('Sum of Even numbers between 100 and 200:', sumEvenNums())     ##  Function sumEvenNums calling
else:
  print('Sum of Odd numbers between 100 and 200:', sumOddNums())


```

```python
def is_palindrome(text):     ## function to check the string is a palindrome or not, if it is palindrome then returns True otherwise False
    processed_text = ''.join(filter(str.isalnum, text)).lower()
    return processed_text == processed_text[::-1]

string1 = "Racecar"


print(f'"{string1}" is a palindrome: {is_palindrome(string1)}')
```
